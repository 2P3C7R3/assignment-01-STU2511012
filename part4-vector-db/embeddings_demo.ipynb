# Embeddings Demo\n\nThis notebook creates sentence embeddings for 10 sentences across exactly 3 topics: **Cricket, Cooking, and Cybersecurity**, computes a cosine similarity matrix, visualizes it as a heatmap, and retrieves the top 2 most similar sentences for a new query.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sentences = [
    "The batsman scored a century in the final match.",
    "The bowler delivered a perfect yorker in the death overs.",
    "The cricket team celebrated after winning the series.",
    "I slow-cooked the curry with onions, garlic, and tomatoes.",
    "Baking bread at home fills the kitchen with a warm aroma.",
    "The chef seasoned the pasta sauce with basil and black pepper.",
    "Strong passwords help prevent unauthorized access to online accounts.",
    "The security team detected malware during the network scan.",
    "Two-factor authentication adds another layer of protection.",
    "Phishing emails often trick users into revealing sensitive data."
]

topics = ["Cricket", "Cricket", "Cricket", "Cooking", "Cooking", "Cooking",
          "Cybersecurity", "Cybersecurity", "Cybersecurity", "Cybersecurity"]

df = pd.DataFrame({"sentence": sentences, "topic": topics})
df
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(sentences)

similarity_matrix = cosine_similarity(embeddings)

sim_df = pd.DataFrame(
    similarity_matrix,
    index=[f"S{i}" for i in range(1, 11)],
    columns=[f"S{i}" for i in range(1, 11)]
)
sim_df.round(2)
```

```python
plt.figure(figsize=(9, 7))
sns.heatmap(sim_df, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title("Cosine Similarity Heatmap for 10 Sentences")
plt.show()
```

```python
query = "The bowler took three wickets in one over"
query_embedding = model.encode([query])
query_similarities = cosine_similarity(query_embedding, embeddings)[0]

top_indices = query_similarities.argsort()[-2:][::-1]

print(f"Query: {query}\n")
for idx in top_indices:
    print(f"Sentence: {sentences[idx]}")
    print(f"Similarity Score: {query_similarities[idx]:.4f}\n")
```
